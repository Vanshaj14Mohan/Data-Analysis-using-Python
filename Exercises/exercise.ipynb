```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales = pd.read_csv('E:\Data Analysis using Python\Exercises\sales_data.csv')
```

```python
sales.head()
```

```python
sales.columns
```

```python
sales["Customer_Age"].describe()
```

```python
sales["Customer_Age"].hist(bins=20)
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.title("Distribution of Customer Ages")
```

# What's the mean of Customers_Age?

```python
sales["Customer_Age"].mean()
```

```python
#Create a KDE plot for the 'Customer_Age' column
sales["Customer_Age"].plot(kind="kde", figsize=(8,6))
plt.xlabel("Age")
plt.ylabel("Density")
plt.legend(["Customer Age Density"])
plt.title("KDE Plot of Customer Ages")
plt.show()
```

```python
# For Box plot
sales["Customer_Age"].plot(kind="box", vert=False, figsize=(6,4))
```

# How many sales do we have per year?

```python
sales.describe()
```

```python
sales["Year"].value_counts()
```

```python
# Creating a pie chart for sales per year
sales["Year"].value_counts().plot(kind="pie", autopct="%1.1f%%", figsize=(10,10))
plt.title("Sales Distribution per Year")
plt.legend(title="Year", loc="best")
plt.show()
```

# How many sales do we have per month?

```python
sales.columns
```

```python
sales["Month"].value_counts()
```

```python
#Plot a bar chart for sales per month
sales["Month"].value_counts().sort_index().plot(kind="bar", figsize=(8,6), color="orange")
plt.xlabel("Month")
plt.ylabel("Number of Sales")
plt.title("Sales per Month")
plt.legend(["Sales Count"])
plt.show()
```

# Which country has the most sales quantity of sales?

```python
sales["Country"].value_counts()
```

```python
sales["Country"].value_counts().plot(kind="bar", figsize=(10,6), color="green")
plt.xlabel("Country")
plt.ylabel("Number of Sales")
plt.title("Sales per Country", color ="blue")
plt.legend(["Sales Count"])
plt.show()
```

# Can you see any relationship between Unit_Cost and Unit_Price?

```python
# Using scatter plot to see realtionship between Unit_Cost and Unit_Price
sales.plot(kind="scatter", x="Unit_Cost", y="Unit_Price", figsize=(8,6), color="red")
plt.xlabel("Unit Cost")
plt.ylabel("Unit Price")
plt.title("Unit Cost vs Unit Price")
plt.legend(["Data Points"], loc="best")
plt.show()
```

# Can you see any relationship between Order_Quantity and Profit?

```python
# Using a scatter plot to see relationship between Unit_Cost and Unit_Price
sales.plot(kind="scatter", x="Unit_Price", y="Profit", figsize=(8,6), color="violet")
plt.xlabel("Unit Price")
plt.ylabel("Profit")
plt.title("Unit Price vs Profit")
plt.legend(["Data Point"], loc="best")
plt.show()
```
