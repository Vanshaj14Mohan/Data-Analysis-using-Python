# Add and calculate a new Calculated_Date column ?

```python
# Use Day, Month, Year to create a Date column (YYYY-MM-DD).
sales["Calculated_Date"] = sales[["Year", "Month", "Day"]].apply(lambda x: '{}-{}-{}'.format(x[0], x[1], x[2]), axis=1)
sales["Calculated_Date"].head()
```

```python
sales.head()
```

```python
sales["Calculated_Date"].describe()
```

# Parse your Calculated_Date column into a datetime object

```python
sales["Calculated_Date"]  =pd.to_datetime(sales["Calculated_Date"])
sales["Calculated_Date"].head()
```

# How did sales evolve through the years?

```python
#Show a line plot using Calculated_Date column as the x-axis and the count of sales as the y-axis.
sales["Calculated_Date"].value_counts().sort_index().plot(kind="line", figsize=(10,8), color="brown")
plt.xlabel("Date")
plt.ylabel("Number of Sales")
plt.title("Number of Sales Over Time")
plt.legend(["Sales Count"])
plt.show()
```
